```python
# A* Search - 8 Puzzle
# Case 2: h(n) = Manhattan Distance
# g(n) = Depth of node
# f(n) = g(n) + h(n)

from heapq import heappush, heappop

# Initial and Goal states
# 0 represents the blank space

initial = (2, 8, 3,
           1, 6, 4,
           0, 7, 5)

goal = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)


# Calculate h(n) using Manhattan Distance
def manhattan_distance(state):
    distance = 0

    for tile in range(1, 9):

        current_position = state.index(tile)
        goal_position = goal.index(tile)

        current_row = current_position // 3
        current_col = current_position % 3

        goal_row = goal_position // 3
        goal_col = goal_position % 3

        distance += abs(current_row - goal_row)
        distance += abs(current_col - goal_col)

    return distance


# Generate possible successor states
def get_neighbors(state):

    neighbors = []

    blank = state.index(0)

    row = blank // 3
    col = blank % 3

    moves = [
        (-1, 0), # Up
        (1, 0), # Down
        (0, -1), # Left
        (0, 1) # Right
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_blank = new_row * 3 + new_col

            new_state = list(state)

            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            neighbors.append(tuple(new_state))

    return neighbors


# Print the puzzle state
def print_state(state):

    for i in range(0, 9, 3):
        print(state[i:i+3])

    print()


# A* Search
def a_star(initial, goal):

    # Priority queue:
    # (f, g, state, path)

    open_list = []

    g_initial = 0
    h_initial = manhattan_distance(initial)
    f_initial = g_initial + h_initial

    heappush(
        open_list,
        (f_initial, g_initial, initial, [initial])
    )

    closed = set()

    while open_list:

        f, g, current, path = heappop(open_list)

        # Goal test
        if current == goal:
            return path

        if current in closed:
            continue

        closed.add(current)

        # Generate successors
        for neighbor in get_neighbors(current):

            if neighbor in closed:
                continue

            new_g = g + 1
            new_h = manhattan_distance(neighbor)
            new_f = new_g + new_h

            heappush(
                open_list,
                (new_f, new_g, neighbor, path + [neighbor])
            )

    return None


# Run A*
solution = a_star(initial, goal)


# Display solution
if solution:

    print("Solution found!")
    print("Number of moves:", len(solution) - 1)
    print()

    for depth, state in enumerate(solution):

        h = manhattan_distance(state)
        f = depth + h

        print("Depth (g) =", depth)
        print("Manhattan Distance (h) =", h)
        print("f =", f)

        print_state(state)

else:

    print("No solution found.")
```

```python
# A* Search - 8 Puzzle
# Case 1: h(n) = Number of Misplaced Tiles
# g(n) = Depth of node
# f(n) = g(n) + h(n)

from heapq import heappush, heappop

# Initial and Goal states
initial = (2, 8, 3,
           1, 6, 4,
           0, 7, 5)       # 0 represents blank

goal = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)


# Calculate h(n)
def misplaced_tiles(state):
    count = 0

    for i in range(9):
        if state[i] != 0 and state[i] != goal[i]:
            count += 1

    return count


# Generate possible successor states
def get_neighbors(state):
    neighbors = []

    blank = state.index(0)
    row = blank // 3
    col = blank % 3

    moves = [
        (-1, 0),   # Up
        (1, 0),    # Down
        (0, -1),   # Left
        (0, 1)     # Right
    ]

    for dr, dc in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_blank = new_row * 3 + new_col

            new_state = list(state)
            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            neighbors.append(tuple(new_state))

    return neighbors


# Print the puzzle
def print_state(state):
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()


# A* Search
def a_star(initial, goal):

    # Priority queue: (f, g, state, path)
    open_list = []

    g_initial = 0
    h_initial = misplaced_tiles(initial)
    f_initial = g_initial + h_initial

    heappush(open_list, (f_initial, g_initial, initial, [initial]))

    closed = set()

    while open_list:

        f, g, current, path = heappop(open_list)

        if current == goal:
            return path

        if current in closed:
            continue

        closed.add(current)

        for neighbor in get_neighbors(current):

            if neighbor in closed:
                continue

            new_g = g + 1
            new_h = misplaced_tiles(neighbor)
            new_f = new_g + new_h

            heappush(
                open_list,
                (new_f, new_g, neighbor, path + [neighbor])
            )

    return None


# Run A*
solution = a_star(initial, goal)


# Display solution
if solution:
    print("Solution found!")
    print("Number of moves:", len(solution) - 1)
    print()

    for depth, state in enumerate(solution):

        h = misplaced_tiles(state)
        f = depth + h

        print("Depth (g) =", depth)
        print("Heuristic (h) =", h)
        print("f =", f)

        print_state(state)

else:
    print("No solution found.")
```
